### Step 2: Load the Dataset into a DataFrame



load the dataset using pd.read_csv()


```python
# Define the URL of the dataset
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3: Identifying Duplicate Rows


**Task 1: Identify Duplicate Rows**
  1. Count the number of duplicate rows in the dataset.
  2. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
# 1) Count the number of duplicate rows in the dataset
num_duplicates = df.duplicated().sum()
print("Number of duplicate rows:", num_duplicates)

# 2) Display the first few duplicate rows to understand their structure
df[df.duplicated(keep=False)].sort_values('ResponseId').head(10)
```

### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


```python
## Write your code here
# 1) Remove duplicate rows from the dataset using drop_duplicates()
df = df.drop_duplicates()

# 2) Verify the removal by counting the number of duplicate rows after removal
print("Duplicate rows remaining:", df.duplicated().sum())
print("Shape after removing duplicates:", df.shape)
```

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


```python
## Write your code here
# 1) Identify missing values for all columns in the dataset
missing = df.isna().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print("Columns with missing values (top 10):")
print(missing.head(10))
print("\nTotal missing values:", df.isna().sum().sum())

# 2) Choose a column with significant missing values (e.g., EdLevel)
#    and impute it with the most frequent value
most_frequent_edlevel = df['EdLevel'].mode()[0]
print("\nMost frequent EdLevel:", most_frequent_edlevel)
print("EdLevel missing before imputation:", df['EdLevel'].isna().sum())
df['EdLevel'] = df['EdLevel'].fillna(most_frequent_edlevel)
print("EdLevel missing after imputation:", df['EdLevel'].isna().sum())
```

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


```python
## Write your code here
# 1) Use the ConvertedCompYearly column for compensation analysis
#    (normalized annual compensation is already provided)
print("ConvertedCompYearly - missing:", df['ConvertedCompYearly'].isna().sum())
print("ConvertedCompYearly - non-null:", df['ConvertedCompYearly'].notna().sum())
print(df['ConvertedCompYearly'].describe())

# 2) Handle missing values: for compensation analysis we keep only respondents
#    who reported compensation (dropping rows without it)
comp_df = df.dropna(subset=['ConvertedCompYearly']).copy()
print("\nCompensation-ready rows:", len(comp_df))
print("Median annual compensation:", comp_df['ConvertedCompYearly'].median())
print("Mean annual compensation:", round(comp_df['ConvertedCompYearly'].mean(), 2))
```
