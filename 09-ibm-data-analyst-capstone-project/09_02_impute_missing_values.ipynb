### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
# Task 1: identify duplicate rows in the dataset
num_duplicates = df.duplicated().sum()
print("Number of duplicate rows:", num_duplicates)
print("Rows involved in duplication (all copies):", df.duplicated(keep=False).sum())

# display the first few duplicate rows to understand their structure
df[df.duplicated(keep=False)].sort_values('ResponseId').head(10)
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
# Task 2: remove the duplicate rows from the dataframe
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
print("Duplicate rows remaining:", df.duplicated().sum())
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
# Task 3: find the missing values for all columns
missing_counts = df.isna().sum().sort_values(ascending=False)
missing_summary = missing_counts[missing_counts > 0]
print("Columns with missing values (", len(missing_summary), "of", df.shape[1], "):")
print(missing_summary.head(10))
print("\nTotal missing values across all columns:", int(missing_counts.sum()))
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
# Task 4: count how many rows are missing in the RemoteWork column
remote_missing = df['RemoteWork'].isna().sum()
print("Missing rows in 'RemoteWork':", remote_missing)
print("Share of rows missing:", round(remote_missing / len(df) * 100, 2), "%")
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
# Task 5: value counts for the RemoteWork column
print(df['RemoteWork'].value_counts())
print("\nTotal rows:", len(df))
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
# Task 6: identify the most frequent (majority) value in RemoteWork
most_frequent_remotework = df['RemoteWork'].mode()[0]
print("Most frequent RemoteWork value:", most_frequent_remotework)
print("Occurrences:", (df['RemoteWork'] == most_frequent_remotework).sum())
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
# Task 7: impute (replace) all empty rows in RemoteWork with the majority value
print("RemoteWork missing before imputation:", df['RemoteWork'].isna().sum())
df['RemoteWork'] = df['RemoteWork'].fillna(most_frequent_remotework)
print("RemoteWork missing after imputation:", df['RemoteWork'].isna().sum())
print("\nUpdated value counts:")
print(df['RemoteWork'].value_counts())
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
# Task 8: check compensation-related columns and describe their distribution
comp_cols = [c for c in df.columns if 'Comp' in c]
print("Compensation-related columns:", comp_cols)
print()
for col in comp_cols:
    print("====", col, "====")
    print("Missing:", df[col].isna().sum(), "of", len(df))
    print(df[col].describe())
    print()
```
